```python
import numpy
```

```python
from transformers import pipeline
```

```python
# List of Hugging Face text generation models
models = [
    "gpt2",
    "distilgpt2",
    "EleutherAI/gpt-neo-125M"
]

# Define the prompt
prompt = "Explain artificial intelligence in simple words."

# Loop through models and generate text
for model in models:
    print(f"\n{'='*20} {model} {'='*20}")
    generator = pipeline("text-generation", model=model)
    output = generator(prompt, max_new_tokens=50, do_sample=True, top_k=50)
    print(output[0]["generated_text"])
```
